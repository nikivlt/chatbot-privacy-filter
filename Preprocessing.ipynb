# Preprocessing of Medical Data

##Patient Records

```python
import pandas as pd
```

The original "patients.csv" includes the personal information for 1162 clinical patients.

```python
patients=pd.read_csv("patients.csv")
patients
```

Because some of the patients are deceased we chose to remove them.

```python
patients = patients [patients ['DEATHDATE'].isna()]
patients
```

Adding a column with the patient's current age.

```python
# Converting birthdate to datetime
patients['BIRTHDATE'] = pd.to_datetime(patients['BIRTHDATE'], errors='coerce')

# Computing age (in years)
today = pd.Timestamp.today()
patients['AGE'] = (today - patients['BIRTHDATE']).dt.days / 365.25
patients['AGE'] = patients['AGE'].astype(int)

# Moving 'AGE' to the third column
cols = patients.columns.tolist()
cols.insert(2, cols.pop(cols.index('AGE')))
patients = patients[cols]
```

The we edit the Martial Status column to better fit natural language.

```python
patients.loc[patients['MARITAL'] == 'M', 'MARITAL'] = 'married'
patients.loc[patients['MARITAL'] == 'S', 'MARITAL'] = 'single'
patients['MARITAL'].fillna("unknown", inplace = True)
patients
```

Finally, we drop some of the columns that contain excessive information.

```python
patients = patients.drop(columns=['BIRTHDATE','DEATHDATE','DRIVERS', 'PASSPORT', 'PREFIX', 'SUFFIX', 'MAIDEN','BIRTHPLACE', 'ETHNICITY', 'COUNTY', 'STATE', 'ZIP', 'LON', 'LAT','HEALTHCARE_EXPENSES', 'HEALTHCARE_COVERAGE' ])
patients
```

## Medical Records

The "medications.csv" contains information about the prescibed medication for some of the patients

```python
md = pd.read_csv("medications.csv")
md
```

We keep only the columns that contain the patients' ids, their prescribed medication and their diagnosis that led to the prescription.

```python
md = md.loc[:, ['PATIENT', 'DESCRIPTION', 'REASONDESCRIPTION']]
md
```

We rename the columns for clarity.

```python
md=md.rename(columns = {"PATIENT":"Id","REASONDESCRIPTION":"DIAGNOSIS", "DESCRIPTION":"MEDICATION"})
```

Because a patient may have multiple perscriptions and diagnoses we group them in a signle row for each patients.

```python
m = md.groupby('Id').agg({'MEDICATION': lambda x: ', '.join(set(x)),'DIAGNOSIS': lambda x: ', '.join(set(x.dropna())) })
m
```

We merge the patients and mediactions tables

```python
patients=pd.merge(patients, m, on = "Id", how = "inner")
patients
```

```python
#We convert the empty strings to NaN
import numpy as np
patients["DIAGNOSIS"] = patients["DIAGNOSIS"].replace("", np.nan)
```

##Allergies

The "allergies.csv" includes information about the patient's allergies

```python
allergies = pd.read_csv("allergies.csv")
allergies = allergies.loc[:, ['PATIENT', 'DESCRIPTION']]
allergies = allergies.rename(columns={'DESCRIPTION':'ALLERGY', 'PATIENT':'Id' })
allergies
```

Because the values int the 'ALLERGY' column are in a ' value(descriptor)' format we chose to remove the descriptor.

```python
import re

allergies['ALLERGY'] = allergies['ALLERGY'].str.replace(r'\(.*?\)', '', regex=True)
```

Then we merge all the allergies for each patient in a single row

```python
allergies = allergies.groupby('Id').agg({'ALLERGY': lambda x: ', '.join(set(x))})
allergies
```

Once again we merge the allergies and patients data frames.

```python
patients = pd.merge(patients, allergies, on = "Id", how = "left")
patients
```

```python
patients = patients.iloc[0:300]
patients
```

```python
patients.to_csv('medical_records.csv')
```

## Conversion to Natural Language

In order to train our csv we must convert our dataframe to natural langauge and export the output in '.docx' format.

```python
#finding out which columns contain NaN values

patients.isna().sum()
```

```python

#The following function returns an empty string if a value is NaN/None/NA.
#Otherwise, it converts the value to a string.
def str_convert(value):

    if pd.isna(value):
        return ""
    return str(value)

#The following funtion converts each row of our patients dataframe to a json object
def row_to_json(row):
    Id = str_convert(row["Id"])
    FIRST = str_convert(row["FIRST"])
    LAST = str_convert(row["LAST"])
    SSN = str_convert(row["SSN"])
    AGE = str_convert(row["AGE"])
    GENDER = str_convert(row["GENDER"])
    MARITAL = str_convert(row["MARITAL"])
    RACE = str_convert(row["RACE"])
    ADDRESS = str_convert(row["ADDRESS"])
    CITY = str_convert(row["CITY"])
    DIAGNOSIS = str_convert(row["DIAGNOSIS"])
    MEDICATION = str_convert(row["MEDICATION"])
    ALLERGY = str_convert(row["ALLERGY"])

    TEXT = (
        f"The patient with id {Id} is named {FIRST} {LAST}. "
        f"Their social security number (SSN) is {SSN}. "
        f"Their age is {AGE} years old, and their gender is listed as {GENDER}. "
        f"The patient's current marital status is {MARITAL}. "
        f"Their race is {RACE} ."
        f"Their home address is {ADDRESS} located in the city of {CITY}. "
    )

    if DIAGNOSIS:
      TEXT += f"As of today, the patient has been diagnosed with: {DIAGNOSIS}. "

    TEXT += f"The medication they have been prescribed includes: {MEDICATION}. "

    if ALLERGY:
         TEXT  += f"Their known allergies are: {ALLERGY}."

    patient_record = {
        "id": Id,
        "name": FIRST,
        "surname": LAST,
        "ssn": SSN,
        "age": AGE,
        "gender": GENDER,
        "marital_status": MARITAL,
        "race": RACE,
        "address": ADDRESS,
        "city": CITY,
        "diagnosis": DIAGNOSIS,
        "medications": MEDICATION,
        "allergies": ALLERGY,
        "text": TEXT

    }

    return patient_record

import json
#each patient record is stored in a .json file
with open("patients.jsonl", "w", encoding="utf-8") as f:
    for _, row in patients.iterrows():
        json.dump(row_to_json(row), f)
        f.write("\n")

```
